```python
from pathlib import Path

folder = Path('inputs')

suffix = 'clean'
for file in folder.iterdir():
    output_file = f'{folder}/{file.stem}_{suffix}.txt'
    with open(file, 'r') as f:
        with open(output_file, 'w') as o:
            lines = f.readlines()
            o.writelines([' '.join((line.split()[1:]+['\n'])) for line in lines])
```
