```python
# Install the required package if you haven't already:
# !pip install langchain-openai

from langchain_openai import ChatOpenAI

# Initialize the model to point to your LM Studio server
llm = ChatOpenAI(
    model="liquid/lfm2.5-1.2b",  # Replace with the exact model name shown in LM Studio
    base_url="http://localhost:1234/v1",  # Default LM Studio API endpoint
    api_key="not-needed",  # type: ignore
    temperature=0,
)

# Test the connection
response = llm.invoke("Hello, who are you?")
print(response.content)
```

```python
from langchain_core.tools import tool

@tool
def get_weather(city: str) -> str:
    """Get the current weather for a city."""
    # In a real app, you'd call a weather API here
    return f"It's sunny in {city}."
```

```python
llm_with_tools = llm.bind_tools([get_weather])

response = llm_with_tools.invoke("What's the weather in Paris?")
```

```python
# The response will contain a .tool_calls attribute if the model decides to use the tool
print(response.tool_calls)
```
